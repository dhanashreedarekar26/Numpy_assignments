Scenario
A retail company has recorded weekly sales (in ₹ thousands) for 4 stores over 12 weeks. The data arrived as a flat 1D array of 48 values with some entries corrupted as np.nan. Your task is to restructure and analyze it.
Tasks
 * Reshape the 1D array into a (4, 12) matrix — rows = stores, columns = weeks.
 * Replace all NaN values with the column (weekly) mean across non-NaN stores.
 * Find the top-performing store (highest total annual sales).
 * Flag weeks where ALL stores sold below ₹50k (boolean mask).

```python
import numpy as np
rng = np.random.default_rng()
sales = rng.integers(1,49,48).astype(np.float64)
ind = np.random.randint(1,48,10)
sales[ind] = np.nan
print(sales)
```

```python
# Reshape
new_sales = sales.reshape(4,12)
print(new_sales)
```

```python
# Replace
mean = np.nanmean(new_sales, axis = 0, keepdims = True)

col = np.isnan(new_sales)

replace = np.where(col, mean, new_sales)
print(replace)
```

```python
# Top performing store
total = np.sum(replace, axis = 1)
print('Top performing store is : ', np.argmax(total)+1,'  : ', np.max(total))
```

```python
# Flag
stores = np.all(replace < 50, axis = 0)
print(stores)
```
